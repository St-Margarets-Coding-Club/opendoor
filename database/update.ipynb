```python
# To access the database
from pymongo.mongo_client import MongoClient
from pymongo.server_api import ServerApi

uri = "mongodb+srv://<dbusername>:<dbpassword>@cluster0.13icold.mongodb.net/?appName=Cluster0"

# Create a new client and connect to the server
client = MongoClient(uri, server_api=ServerApi('1'))

# Send a ping to confirm a successful connection
try:
    client.admin.command('ping')
    print("Pinged your deployment. You successfully connected to MongoDB!")
except Exception as e:
    print(e)
```

```python
# Access the database
db = client['animalsdb']
```

```python
# Access the collection
collection = db['animals']
```

```python
# Create a new field in every document
collection.update_many({}, {"$set": {"images": None}})
```

```python
# Update field name of images to image
collection.update_many({}, {"$rename": {"images": "image"}})
```
